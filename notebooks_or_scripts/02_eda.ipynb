```python
#load
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
train=pd.read_csv('../data/mitbih_train.csv',header=None)
test=pd.read_csv('../data/mitbih_test.csv',header=None)
print(train.shape)
print(test.shape)
```

```python
#class balance
names={0: 'Normal(N)', 1: 'Supraventricular(S)', 2: 'Ventricular(V)', 3: 'Fusion(F)', 4: 'Unknown(Q)'}
counts=train[187].value_counts().sort_index()
print(counts)
print(counts.rename(index=names))
```

```python

print("percentages of each class in training set")
print((counts/counts.sum()*100).round(1).rename(index=names))
counts.rename(index=names).plot(kind='bar',rot=20,title='Training set classes counts')
plt.tight_layout()
plt.show()
```

```python
#one heartbeat plot for each class
fig,axes=plt.subplots(5,1,figsize=(10,12),sharex=True)
for label,ax in zip(range(5),axes):
    row=train[train[187]==label].iloc[0,:187]
    ax.plot(row.values)
    ax.set_title(names[label])
axes[-1].set_xlabel('Time step')
plt.tight_layout()
plt.show()
```

```python
#sanity checks
print("Missing values:",train.isna().sum().sum())
print("min and max signal values:",train.iloc[:,:187].values.min(),train.iloc[:,:187].values.max())
```
